```python
# LAB ASSIGNMENT 3 - AI IN HEALTHCARE

!pip install -q pandas numpy matplotlib seaborn scipy scikit-learn
!pip install -q nltk tensorflow wfdb opencv-python kagglehub



import os
import re
import zipfile
import urllib.request

import cv2
import wfdb
import nltk
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from collections import Counter

from scipy import stats
from scipy.signal import butter, filtfilt

from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.feature_extraction.text import TfidfVectorizer

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator



BASE_DIR = os.getcwd()

DATA_DIR = os.path.join(BASE_DIR, "datasets")

PIMA_DIR = os.path.join(
    DATA_DIR,
    "pima_diabetes"
)

MTSAMPLES_DIR = os.path.join(
    DATA_DIR,
    "mtsamples"
)

XRAY_DIR = os.path.join(
    DATA_DIR,
    "chest_xray"
)

ECG_DIR = os.path.join(
    DATA_DIR,
    "mit_bih_ecg"
)


# Create directories automatically

os.makedirs(
    PIMA_DIR,
    exist_ok=True
)

os.makedirs(
    MTSAMPLES_DIR,
    exist_ok=True
)

os.makedirs(
    XRAY_DIR,
    exist_ok=True
)

os.makedirs(
    ECG_DIR,
    exist_ok=True
)


print("PROJECT DIRECTORY:")
print(BASE_DIR)

print("\nDATA DIRECTORY:")
print(DATA_DIR)



nltk.download("stopwords")
nltk.download("punkt")

from nltk.corpus import stopwords


print("\n" + "=" * 70)
print("PART A - PIMA INDIANS DIABETES DATASET")
print("=" * 70)



PIMA_FILE = os.path.join(
    PIMA_DIR,
    "diabetes.csv"
)


if not os.path.exists(PIMA_FILE):

    print("\nDownloading Pima Diabetes Dataset...")

    pima_url = (
        "https://raw.githubusercontent.com/"
        "plotly/datasets/master/diabetes.csv"
    )

    try:

        urllib.request.urlretrieve(
            pima_url,
            PIMA_FILE
        )

        print(
            "Pima dataset downloaded successfully!"
        )

    except Exception as e:

        print(
            "Could not download Pima dataset:"
        )

        print(e)


else:

    print(
        "\nPima dataset already exists."
    )



df = pd.read_csv(
    PIMA_FILE
)


print("\nDataset Path:")
print(PIMA_FILE)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)



print("\nMissing Values:")
print(df.isnull().sum())


columns_with_invalid_zero = [

    "Glucose",

    "BloodPressure",

    "SkinThickness",

    "Insulin",

    "BMI"

]


df[
    columns_with_invalid_zero
] = df[
    columns_with_invalid_zero
].replace(
    0,
    np.nan
)


# Fill missing values with median

for column in columns_with_invalid_zero:

    df[column] = df[column].fillna(

        df[column].median()

    )


print("\nMissing Values After Cleaning:")
print(df.isnull().sum())



print("\nStatistical Summary:")
print(df.describe())


df.hist(
    figsize=(15, 10)
)

plt.suptitle(
    "Pima Diabetes Dataset"
)

plt.tight_layout()

plt.show()


# Correlation Heatmap

plt.figure(
    figsize=(10, 8)
)

sns.heatmap(

    df.corr(),

    annot=True,

    cmap="coolwarm"

)

plt.title(
    "Pima Diabetes Correlation Heatmap"
)

plt.show()


# Outcome Distribution

plt.figure(
    figsize=(6, 5)
)

sns.countplot(

    x="Outcome",

    data=df

)

plt.title(
    "Diabetes Outcome Distribution"
)

plt.show()


X = df.drop(
    "Outcome",
    axis=1
)

y = df[
    "Outcome"
]


scaler = StandardScaler()


X_scaled = scaler.fit_transform(
    X
)


print(
    "\nFeature Scaling Completed"
)



diabetic_glucose = df[
    df["Outcome"] == 1
]["Glucose"]


non_diabetic_glucose = df[
    df["Outcome"] == 0
]["Glucose"]


t_statistic, p_value = stats.ttest_ind(

    diabetic_glucose,

    non_diabetic_glucose

)


print("\nT TEST RESULT")

print(
    "T Statistic:",
    t_statistic
)

print(
    "P Value:",
    p_value
)


if p_value < 0.05:

    print(
        "Reject Null Hypothesis"
    )

    print(
        "Glucose levels differ significantly."
    )

else:

    print(
        "Fail to Reject Null Hypothesis"
    )



df["AgeGroup"] = pd.cut(

    df["Age"],

    bins=[
        20,
        30,
        40,
        50,
        100
    ],

    labels=[
        "20-30",
        "31-40",
        "41-50",
        "50+"
    ]

)


contingency_table = pd.crosstab(

    df["AgeGroup"],

    df["Outcome"]

)


print("\nCHI SQUARE TABLE")

print(
    contingency_table
)


chi2, p_chi, dof, expected = stats.chi2_contingency(

    contingency_table

)


print(
    "\nChi Square Statistic:",
    chi2
)

print(
    "P Value:",
    p_chi
)



group1 = df[
    df["AgeGroup"] == "20-30"
]["BMI"]

group2 = df[
    df["AgeGroup"] == "31-40"
]["BMI"]

group3 = df[
    df["AgeGroup"] == "41-50"
]["BMI"]

group4 = df[
    df["AgeGroup"] == "50+"
]["BMI"]


f_statistic, p_anova = stats.f_oneway(

    group1,

    group2,

    group3,

    group4

)


print("\nANOVA RESULT")

print(
    "F Statistic:",
    f_statistic
)

print(
    "P Value:",
    p_anova
)


X_feature = df.drop(

    columns=[
        "Outcome",
        "AgeGroup"
    ]

)


selector = SelectKBest(

    score_func=f_classif,

    k=5

)


X_selected = selector.fit_transform(

    X_feature,

    y

)


selected_features = X_feature.columns[

    selector.get_support()

]


print("\nTOP 5 SELECTED FEATURES:")

print(
    selected_features.tolist()
)



print("\n" + "=" * 70)
print("PART B - MTSAMPLES MEDICAL TEXT DATASET")
print("=" * 70)



MTSAMPLES_FILE = os.path.join(

    MTSAMPLES_DIR,

    "mtsamples.csv"

)



if not os.path.exists(MTSAMPLES_FILE):

    print("\nDownloading MTSamples Dataset...")


    mtsamples_url = (

        "https://raw.githubusercontent.com/"
        "Midhilesh489/Medical-Text-Dataset/"
        "main/mtsamples.csv"

    )


    try:

        urllib.request.urlretrieve(

            mtsamples_url,

            MTSAMPLES_FILE

        )


        print(
            "MTSamples downloaded successfully!"
        )


    except Exception as e:

        print(
            "Automatic download failed."
        )

        print(e)

        print(
            "\nPlease manually place mtsamples.csv at:"
        )

        print(
            MTSAMPLES_FILE
        )



if os.path.exists(
    MTSAMPLES_FILE
):

    df_text = pd.read_csv(

        MTSAMPLES_FILE

    )


    print(
        "\nDataset Path:"
    )

    print(
        MTSAMPLES_FILE
    )


    print(
        "\nDataset Columns:"
    )

    print(
        df_text.columns
    )


    print(
        "\nFirst 5 Rows:"
    )

    print(
        df_text.head()
    )


    df_text = df_text.dropna(

        subset=[
            "transcription"
        ]

    )



    stop_words = set(

        stopwords.words(
            "english"
        )

    )


    def clean_text(text):

        text = str(text).lower()


        text = re.sub(

            r"[^a-zA-Z\s]",

            "",

            text

        )


        words = text.split()


        words = [

            word

            for word in words

            if word not in stop_words

        ]


        return " ".join(
            words
        )


    df_text["clean_text"] = df_text[

        "transcription"

    ].apply(
        clean_text
    )


    print(
        "\nText Cleaning Completed"
    )



    all_text = " ".join(

        df_text[
            "clean_text"
        ]

    )


    word_counts = Counter(

        all_text.split()

    )


    common_words = word_counts.most_common(
        15
    )


    words = [

        item[0]

        for item in common_words

    ]


    counts = [

        item[1]

        for item in common_words

    ]


    plt.figure(
        figsize=(12, 6)
    )


    plt.bar(
        words,
        counts
    )


    plt.xticks(
        rotation=45
    )


    plt.title(
        "Top Medical Words"
    )


    plt.tight_layout()

    plt.show()




    vectorizer = TfidfVectorizer(

        max_features=5000

    )


    X_text = vectorizer.fit_transform(

        df_text[
            "clean_text"
        ]

    )


    print(
        "\nTF-IDF Shape:"
    )

    print(
        X_text.shape
    )


else:

    print(
        "\nMTSamples Dataset Not Available"
    )


print("\n" + "=" * 70)
print("PART C - CHEST X-RAY PNEUMONIA DATASET")
print("=" * 70)




try:

    import kagglehub


    print(
        "\nDownloading Chest X-Ray Dataset..."
    )


    dataset_path = kagglehub.dataset_download(

        "paultimothymooney/chest-xray-pneumonia"

    )


    print(
        "\nKaggle Dataset Downloaded!"
    )


    print(
        "Dataset Path:"
    )

    print(
        dataset_path
    )


    # Find chest_xray directory

    possible_path = os.path.join(

        dataset_path,

        "chest_xray"

    )


    if os.path.exists(
        possible_path
    ):

        XRAY_DATASET_PATH = possible_path

    else:

        XRAY_DATASET_PATH = dataset_path


    TRAIN_PATH = os.path.join(

        XRAY_DATASET_PATH,

        "train"

    )


    TEST_PATH = os.path.join(

        XRAY_DATASET_PATH,

        "test"

    )


    print(
        "\nTraining Path:"
    )

    print(
        TRAIN_PATH
    )


    print(
        "\nTesting Path:"
    )

    print(
        TEST_PATH
    )




    train_datagen = ImageDataGenerator(

        rescale=1.0 / 255,

        rotation_range=10,

        width_shift_range=0.1,

        height_shift_range=0.1,

        zoom_range=0.1,

        horizontal_flip=True

    )


    test_datagen = ImageDataGenerator(

        rescale=1.0 / 255

    )



    train_data = train_datagen.flow_from_directory(

        TRAIN_PATH,

        target_size=(
            150,
            150
        ),

        batch_size=32,

        class_mode="binary"

    )




    test_data = test_datagen.flow_from_directory(

        TEST_PATH,

        target_size=(
            150,
            150
        ),

        batch_size=32,

        class_mode="binary"

    )


    print(
        "\nClass Mapping:"
    )

    print(
        train_data.class_indices
    )


    images, labels = next(
        train_data
    )


    plt.figure(
        figsize=(10, 10)
    )


    for i in range(9):

        plt.subplot(
            3,
            3,
            i + 1
        )


        plt.imshow(
            images[i]
        )


        if labels[i] == 0:

            plt.title(
                "NORMAL"
            )

        else:

            plt.title(
                "PNEUMONIA"
            )


        plt.axis(
            "off"
        )


    plt.tight_layout()

    plt.show()



    model = tf.keras.Sequential([


        tf.keras.layers.Conv2D(

            32,

            (3, 3),

            activation="relu",

            input_shape=(
                150,
                150,
                3
            )

        ),


        tf.keras.layers.MaxPooling2D(),


        tf.keras.layers.Conv2D(

            64,

            (3, 3),

            activation="relu"

        ),


        tf.keras.layers.MaxPooling2D(),


        tf.keras.layers.Conv2D(

            128,

            (3, 3),

            activation="relu"

        ),


        tf.keras.layers.MaxPooling2D(),


        tf.keras.layers.Flatten(),


        tf.keras.layers.Dense(

            128,

            activation="relu"

        ),


        tf.keras.layers.Dropout(
            0.5
        ),


        tf.keras.layers.Dense(

            1,

            activation="sigmoid"

        )

    ])



    model.compile(

        optimizer="adam",

        loss="binary_crossentropy",

        metrics=[
            "accuracy"
        ]

    )


    model.summary()




    history = model.fit(

        train_data,

        validation_data=test_data,

        epochs=3

    )




    plt.figure(
        figsize=(8, 5)
    )


    plt.plot(

        history.history[
            "accuracy"
        ],

        label="Training Accuracy"

    )


    plt.plot(

        history.history[
            "val_accuracy"
        ],

        label="Validation Accuracy"

    )


    plt.legend()


    plt.title(
        "CNN Accuracy"
    )


    plt.show()


except Exception as e:

    print(
        "\nChest X-Ray Dataset Error:"
    )

    print(e)

    print(
        "\nIf Kaggle authentication is required:"
    )

    print(
        "Upload your kaggle.json file in Google Colab."
    )



print("\n" + "=" * 70)
print("PART D - MIT-BIH ARRHYTHMIA ECG DATASET")
print("=" * 70)


try:

    print(
        "\nDownloading MIT-BIH ECG Record 100..."
    )


    record = wfdb.rdrecord(

        "100",

        pn_dir="mitdb"

    )


    signal = record.p_signal


    print(
        "\nECG Signal Loaded Successfully"
    )


    print(
        "Signal Shape:",
        signal.shape
    )


    # SAVE ECG LOCALLY

    np.save(

        os.path.join(

            ECG_DIR,

            "ecg_signal_100.npy"

        ),

        signal

    )


    print(
        "\nECG saved at:"
    )

    print(
        ECG_DIR
    )



    # PLOT ORIGINAL SIGNAL

    plt.figure(
        figsize=(15, 5)
    )


    plt.plot(

        signal[
            :2000,
            0
        ]

    )


    plt.title(
        "Original ECG Signal"
    )


    plt.xlabel(
        "Samples"
    )


    plt.ylabel(
        "Amplitude"
    )


    plt.show()


    # BUTTERWORTH FILTER

    def butter_lowpass_filter(

        data,

        cutoff,

        fs,

        order=5

    ):


        nyquist = 0.5 * fs


        normalized_cutoff = (

            cutoff / nyquist

        )


        b, a = butter(

            order,

            normalized_cutoff,

            btype="low"

        )


        filtered_data = filtfilt(

            b,

            a,

            data

        )


        return filtered_data


    # Sampling frequency of MIT-BIH

    fs = 360


    filtered_signal = butter_lowpass_filter(

        signal[:, 0],

        cutoff=40,

        fs=fs

    )


    #
    # COMPARE SIGNALS
    #

    plt.figure(
        figsize=(15, 5)
    )


    plt.plot(

        signal[
            :2000,
            0
        ],

        label="Original"

    )


    plt.plot(

        filtered_signal[
            :2000
        ],

        label="Filtered"

    )


    plt.legend()


    plt.title(
        "ECG Noise Removal"
    )


    plt.show()


    # SIGNAL NORMALIZATION

    signal_scaler = StandardScaler()


    normalized_signal = signal_scaler.fit_transform(

        filtered_signal.reshape(
            -1,
            1
        )

    )


    print(
        "\nNormalized Signal Shape:"
    )

    print(
        normalized_signal.shape
    )


    # SIGNAL WINDOWING

    window_size = 500


    windows = []


    for i in range(

        0,

        len(filtered_signal) - window_size,

        window_size

    ):


        window = filtered_signal[

            i:
            i + window_size

        ]


        windows.append(
            window
        )


    windows = np.array(
        windows
    )


    print(
        "\nWindowed Signal Shape:"
    )

    print(
        windows.shape
    )


except Exception as e:

    print(
        "\nECG ERROR:"
    )

    print(e)


# FINAL OUTPUT

print("\n" + "=" * 70)

print(
    "LAB ASSIGNMENT 3 COMPLETED"
)

print("=" * 70)


print(
    "\nALL DATASET DIRECTORIES:"
)

print(
    "\nPima:"
)

print(
    PIMA_DIR
)


print(
    "\nMTSamples:"
)

print(
    MTSAMPLES_DIR
)


print(
    "\nChest X-Ray:"
)

print(
    XRAY_DIR
)


print(
    "\nECG:"
)

print(
    ECG_DIR
)


print("\n" + "=" * 70)
print("END OF PROGRAM")
print("=" * 70)
```
