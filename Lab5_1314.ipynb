```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report,
                             roc_curve, auc, roc_auc_score)
from itertools import cycle

# ==========================================
# Task 1: Data Acquisition and Exploration
# ==========================================

# 1. Download and load the dataset
# Note: Ensure 'dermatology.csv' is in your working directory.
# The Dermatology dataset has 'Class' as the target variable (1-6)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/dermatology/dermatology.data"
columns = [f"feature_{i}" for i in range(1, 35)] + ['Class']
df = pd.read_csv(url, names=columns, na_values='?')

# 2. Compute summary statistics
print("--- Summary Statistics ---")
display(df.describe())

# 3. Check for missing values, duplicates, or outliers and handle them
print("\nMissing values per column:\n", df.isnull().sum())
print("Total duplicate rows:", df.duplicated().sum())

# Handle missing values (impute with median for numerical features)
df.fillna(df.median(), inplace=True)
# Drop duplicates
df.drop_duplicates(inplace=True)

# 4. Visualize the data
plt.figure(figsize=(10, 6))
sns.countplot(x='Class', data=df, palette='viridis')
plt.title('Class Distribution (1: Psoriasis, 2: Seboreic Dermatitis, etc.)')
plt.show()

# Histograms for first 5 features to avoid clutter
df.iloc[:, :5].hist(bins=15, figsize=(10, 8), layout=(2, 3))
plt.suptitle('Histograms of First 5 Features')
plt.show()

# Correlation heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()

# 5. Discuss imbalances (Print statement placeholder for report)
print("\n--- Imbalance Discussion ---")
print("If the countplot shows significant disparity between class counts (e.g., Class 1 has 100 samples and Class 6 has 20), the dataset is imbalanced. This can cause the model to be biased toward the majority class, leading to poor recall for minority classes. Stratified sampling and evaluation metrics like F1-score (macro) are essential here.")

# ==========================================
# Task 2: Data Preprocessing
# ==========================================

X = df.drop('Class', axis=1)
y = df['Class']

# Encode the target variable (Standardizes classes to start from 0 for ML algorithms)
le = LabelEncoder()
y_encoded = le.fit_transform(y)
classes = le.classes_

# Split data (80/20 split) with stratified sampling
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==========================================
# Task 3: Model Training and Basic Evaluation
# ==========================================

# Initialize models
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Support Vector Machine": SVC(probability=True, random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5)
}

# Binarize the output for ROC curve calculations (Multiclass OvR)
y_test_bin = label_binarize(y_test, classes=range(len(classes)))
n_classes = y_test_bin.shape[1]

results = {}

for name, model in models.items():
    print(f"\n{'='*40}\nEvaluating Model: {name}\n{'='*40}")

    # Train
    model.fit(X_train_scaled, y_train)

    # Predict
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)

    # Metrics
    acc = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {acc:.4f}")
    print("\nClassification Report (Precision, Recall, F1-Score):")
    print(classification_report(y_test, y_pred, target_names=[str(c) for c in classes]))

    # Confusion Matrix Visualization
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
    plt.title(f'Confusion Matrix: {name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

    # ROC Curve and AUC Score
    roc_auc = roc_auc_score(y_test, y_prob, multi_class='ovr')
    print(f"ROC AUC Score (Macro OvR): {roc_auc:.4f}")

    # Plot Multiclass ROC
    fpr = dict()
    tpr = dict()
    roc_auc_dict = dict()
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
        roc_auc_dict[i] = auc(fpr[i], tpr[i])

    plt.figure(figsize=(8, 6))
    colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red', 'purple'])
    for i, color in zip(range(n_classes), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2,
                 label=f'ROC curve of class {classes[i]} (area = {roc_auc_dict[i]:.2f})')
    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'Multiclass ROC Curve - {name}')
    plt.legend(loc="lower right", fontsize='small')
    plt.show()

print("\n--- Model Interpretation ---")
print("1. Compare the Accuracy and F1-scores across models to determine the best baseline performer.")
print("2. The confusion matrix highlights which specific skin diseases are frequently misclassified (e.g., confusing Class 1 with Class 2).")
print("3. High AUC scores (close to 1.0) on the ROC curves indicate strong capability of the model to separate that specific class from the rest.")
```
